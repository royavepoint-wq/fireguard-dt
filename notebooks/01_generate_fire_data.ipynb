# 01 - Synthetic Training Dataset Generation
This notebook creates the **Synthetic Training Dataset** for Step 5 fire-risk classification.

Random seed: `42`

```python
from pathlib import Path
import importlib.util

project_root = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
module_path = project_root / 'backend' / 'app' / 'ml_pipeline' / 'synthetic_data.py'
spec = importlib.util.spec_from_file_location('synthetic_data', module_path)
module = importlib.util.module_from_spec(spec)
assert spec is not None and spec.loader is not None
spec.loader.exec_module(module)

RANDOM_STATE = module.RANDOM_STATE
TRAINING_ROWS = module.TRAINING_ROWS
generate_synthetic_fire_dataset = module.generate_synthetic_fire_dataset

dataset = generate_synthetic_fire_dataset(rows=TRAINING_ROWS, random_state=RANDOM_STATE)
out_path = project_root / 'data' / 'ml' / 'fire_sensor_dataset.csv'
out_path.parent.mkdir(parents=True, exist_ok=True)
dataset.to_csv(out_path, index=False)
print(f'Saved: {out_path}')
```

```python
print('Shape:', dataset.shape)
display(dataset.head())
display(dataset['risk_class'].value_counts(normalize=True).rename('ratio'))
display(dataset.isna().sum())
display(dataset.describe(include='all'))
```
